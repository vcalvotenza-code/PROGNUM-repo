```python
class Fibonacci:
    def __init__(self):
        self.first = 0
        self.second = 1
        
    def Nth_term(self, N):
        a = self.first
        b = self.second
        
        for i in range(N-1):
            a, b= b , a+b
        return a
    def Divisible(self, N, M):
        results=[]
        
        a=self.first
        b=self.second
        
        for i in range(N-1):
            if a % M ==0:
                results.append(a)
            a, b = b, a+b
        return results
    
fib = Fibonacci()

N=100
M=7

print(f'the {N}th term is: {fib.Nth_term(N)} and the fibonacci numbers less than {N}th term divisible by {M} are:{fib.Divisible(N, M)}')
```
